```python

X = [75, 65, 80]
w = [4, 3, 5]
b = -1000
eta = 0.01
y = 1

print("Initial Weights:", w)
print("Initial Bias:", b)

z = (w[0]*X[0]) + (w[1]*X[1]) + (w[2]*X[2]) + b
print("\nNet input (z):", z)

if z >= 0:
    y_pred = 1
else:
    y_pred = 0

print("Predicted Output:", y_pred)
error = y - y_pred
print("Error:", error)

if error != 0:
    for i in range(len(w)):
        w[i] = w[i] + eta * error * X[i]
    b = b + eta * error

print("\nUpdated Weights:", w)
print("Updated Bias:", b)

z_new = (w[0]*X[0]) + (w[1]*X[1]) + (w[2]*X[2]) + b

if z_new >= 0:
    y_new = 1
else:
    y_new = 0

print("\nAfter Learning:")
print("New z:", z_new)
print("New Prediction:", y_new)
```
